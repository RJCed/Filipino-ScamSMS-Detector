```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
```

```python
data = pd.read_csv('dataset.csv')

mess = data['message']
label = data['label']
```

```python
(mess_train, mess_test, label_train, label_test) = train_test_split(mess, label, test_size=0.2)

cv = CountVectorizer(lowercase=True, stop_words=None, ngram_range=(1, 2))
features = cv.fit_transform(mess_train)
```

```python
model = MultinomialNB()
model.fit(features, label_train)
```

```python
features_test = cv.transform(mess_test)
```

```python
message = cv.transform([""]).toarray()
result = model.predict(message)
print(result)
```
